## 1. 创建 Tensor

`torch.tensor(...)` 可以从 Python 的 list 直接建一个 tensor。此外还有一批「工厂函数」：

- `zeros` —— 全 0
- `ones` —— 全 1
- `randn` —— 标准正态分布随机数，以后初始化网络权重经常用它
- `arange` —— 等差数列，类似 Python 的 `range`

````python
a = torch.tensor([1, 2, 3])
print('从 list 创建 :', a)

print('全 0  (2x3) :', torch.zeros(2, 3))
print('全 1  (2x3) :', torch.ones(2, 3))

# randn = 标准正态分布随机数
print('随机  (2x3) :', torch.randn(2, 3))

# arange(起点, 终点, 步长) —— 注意【不含】终点
print('等差 0,2,4,6,8 :', torch.arange(0, 10, 2))
````

## 2. shape / dtype / device —— tensor 的「三围」

这三个属性你会看一辈子，几乎每次 debug 都在看它们：

- `shape` —— 形状。`(3, 4)` 表示 3 行 4 列。**矩阵乘能不能做，全看它。**
- `dtype` —— 数据类型。深度学习里基本是 `float32`。
- `device` —— 在 CPU 还是 GPU 上。**最坑人的报错就是两个 tensor 设备不匹配。**

````python
x = torch.randn(3, 4)
print('x        :', x)
print('shape    :', x.shape)      # 形状
print('dtype    :', x.dtype)      # 元素类型
print('device   :', x.device)     # 在哪个设备上
print('ndim     :', x.ndim)       # 维度数
print('numel    :', x.numel())    # 元素总个数
````

````python
# 轮到你 2
# 创建一个 shape 为 (2, 5) 的随机 tensor，然后分别打印它的
# shape、dtype、numel()
#
# 提示：torch.randn(2, 5)
y=torch.randn(2, 5)
print('y        :', y.shape)
print('dtype    :', y.dtype)
print('numel    :', y.numel())
````

## 3. reshape —— 改形状，不改数据

`reshape` 只重新排列元素的「读法」，元素本身和顺序都不变。12 个元素可以看成 `(12,)`、`(3, 4)`、`(2, 6)`、`(4, 3)`…… **元素总数必须对得上**，否则报错。

> 为什么重要：图像从 dataloader 出来常常是 `(batch, 3, 32, 32)`，进去模型前经常要「拍平」成 `(batch, 3*32*32)`。

````python
x = torch.arange(12)
print('原始 :', x, '->', x.shape)

y = x.reshape(3, 4)
print('reshape(3, 4):')
print(y, '->', y.shape)

z = x.reshape(2, 6)
print('reshape(2, 6) ->', z.shape)

# 用 -1 让 PyTorch 自己推断那一维
w = x.reshape(4, -1)
print('reshape(4, -1) ->', w.shape)
````

## 4. 索引和切片

和 Python list 的 `x[i]`、`x[a:b]` 一个道理，只是多了「多维」：`x[行, 列]`。冒号 `:` 表示「这一维全都要」。

````python
x = torch.arange(12).reshape(3, 4)
print('x =')
print(x)
print()
print('x[0]      :', x[0])        # 第 0 行
print('x[1]      :', x[1])        # 第 1 行
print('x[0, 1]   :', x[0, 1])     # 第 0 行 第 1 列
print('x[:, 0]   :', x[:, 0])     # 所有行 的第 0 列
print('x[1:, :2] :')
print(x[1:, :2])                   # 第 1 行起，前 2 列
````

````python
# 轮到你 4
# 用上面的 x，分别取出：
#   (a) 最后一行
#   (b) 第 2 列（所有行）
#   (c) 右下角那个元素
#
# 提示：负数索引 x[-1] 也可以用
import torch
x = torch.arange(12).reshape(3, 4)
print('x =', x)
print('x 最后一行：',x[-1])
print('x 第 2 列：',x[:, 1])
print('x 右下角元素：',x[-1, -1])
````

## 5. 矩阵乘法 —— 今天的重头戏

这是你以后在 Transformer 里天天见的东西。规则只有一条：

```
(3, 4) @ (4, 5)  ->  (3, 5)
        ^  ^
    这两个数必须相等
```

**中间两个数必须一样，结果取两头的数** —— 记住这一句就够了。

两种写法完全等价：`A @ B` 和 `torch.matmul(A, B)`。

````python
A = torch.randn(3, 4)
B = torch.randn(4, 5)
C = A @ B
print(A.shape, '@', B.shape, '->', C.shape)

D = torch.matmul(A, B)   # 等价写法
print('matmul 结果一样吗？', torch.allclose(C, D))
````

## 6. CPU / GPU

默认所有 tensor 都在 CPU 上。要上 GPU 得显式搬过去：`x.to('cuda')`。

**规则**：两个 tensor 要做运算，必须在同一个设备上。一个 `cuda`、一个 `cpu` 直接相加会报错 —— 这是新手最常见的错误之一。

````python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('这台机器用:', device)

x = torch.randn(3, 4)
print('创建时   :', x.device)

x = x.to(device)
print('搬过去后 :', x.device)

if device == 'cuda':
    print('显卡     :', torch.cuda.get_device_name(0))
    print('显存(GB) :', round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 1))
````

````python
# 轮到你 6
# (a) 建两个 (4, 4) 的随机 tensor，都搬到 device 上，相乘，打印结果的 device
# (b) 再故意把其中一个留在 CPU，两个相乘 —— 看报错长什么样
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
x = torch.randn(4, 4)
y = torch.randn(4, 4)
print ('x是：',x)
print ('y是：',y)

x = x.to(device)
y = y.to(device)
z = x @ y
print ("x@y 结果", z)
````
